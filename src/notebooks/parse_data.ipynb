```python
from pathlib import Path
import pandas as pd
```

```python
data_dir = Path("../data")
bronze_dir = data_dir / "bronze"
silver_dir = data_dir / "silver"
```

```python
unlabled_columns = [
    "duration",
    "protocol_type",
    "service",
    "flag",
    "src_bytes",
    "dst_bytes",
    "land",
    "wrong_fragment",
    "urgent",
    "hot",
    "num_failed_logins",
    "logged_in",
    "num_compromised",
    "root_shell",
    "su_attempted",
    "num_root",
    "num_file_creations",
    "num_shells",
    "num_access_files",
    "num_outbound_cmds",
    "is_host_login",
    "is_guest_login",
    "count",
    "srv_count",
    "serror_rate",
    "srv_serror_rate",
    "rerror_rate",
    "srv_rerror_rate",
    "same_srv_rate",
    "diff_srv_rate",
    "srv_diff_host_rate",
    "dst_host_count",
    "dst_host_srv_count",
    "dst_host_same_srv_rate",
    "dst_host_diff_srv_rate",
    "dst_host_same_src_port_rate",
    "dst_host_srv_diff_host_rate",
    "dst_host_serror_rate",
    "dst_host_srv_serror_rate",
    "dst_host_rerror_rate",
    "dst_host_srv_rerror_rate",
]
```

```python
columns = [*unlabled_columns, "attack_type"]
```

```python
test_data = pd.read_csv(bronze_dir / "corrected.csv", names=columns)
print(f"Corrected data shape: {test_data.shape}")
test_data.head()
```

```python
train_data = pd.read_csv(bronze_dir / "kddcup_data_10_percent.csv", names=columns)
print(f"Training data shape: {train_data.shape}")
train_data.head()
```

```python
unlabeled_test_data = pd.read_csv(bronze_dir / "kddcup_testdata_unlabeled_10_percent.csv", names=unlabled_columns)
```
