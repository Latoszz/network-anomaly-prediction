# Transform data

import libraries

```python
from pathlib import Path
import pandas as pd
import numpy as np
```

set variables

```python
data_dir = Path("../data")
bronze_dir = data_dir / "bronze"
silver_dir = data_dir / "silver"
```

```python
attack_map = {
    'normal.': 'normal',
    
    # DoS
    'back.': 'dos',
    'land.': 'dos',
    'neptune.': 'dos',
    'pod.': 'dos',
    'smurf.': 'dos',
    'teardrop.': 'dos',
    'mailbomb.': 'dos',
    'processtable.':'dos',
    'udpstorm.':'dos',

    # Probe
    'ipsweep.': 'probe',
    'nmap.': 'probe',
    'portsweep.': 'probe',
    'satan.': 'probe',
    'snmpgetattack.': 'probe',
    'snmpguess.': 'probe',
    'mscan.': 'probe',
    'saint.': 'probe',
    'ps.': 'probe',
    
    # R2L
    'ftp_write.': 'r2l',
    'guess_passwd.': 'r2l',
    'imap.': 'r2l',
    'multihop.': 'r2l',
    'phf.': 'r2l',
    'spy.': 'r2l',
    'warezclient.': 'r2l',
    'warezmaster.': 'r2l',
    'apache2.': 'r2l',
    'httptunnel.': 'r2l',
    'named.': 'r2l',
    'worm.': 'r2l',
    'sqlattack.': 'r2l',
    
    # U2R
    'buffer_overflow.': 'u2r',
    'loadmodule.': 'u2r',
    'perl.': 'u2r',
    'rootkit.': 'u2r',
    'sendmail.': 'u2r',
    'xterm.': 'u2r',
    'xlock.': 'u2r',
    'xsnoop.': 'u2r',
}
```

```python
columns = [
    "duration",
    "protocol_type",
    "service",
    "flag",
    "src_bytes",
    "dst_bytes",
    "land",
    "wrong_fragment",
    "urgent",
    "hot",
    "num_failed_logins",
    "logged_in",
    "num_compromised",
    "root_shell",
    "su_attempted",
    "num_root",
    "num_file_creations",
    "num_shells",
    "num_access_files",
    "num_outbound_cmds",
    "is_host_login",
    "is_guest_login",
    "count",
    "srv_count",
    "serror_rate",
    "srv_serror_rate",
    "rerror_rate",
    "srv_rerror_rate",
    "same_srv_rate",
    "diff_srv_rate",
    "srv_diff_host_rate",
    "dst_host_count",
    "dst_host_srv_count",
    "dst_host_same_srv_rate",
    "dst_host_diff_srv_rate",
    "dst_host_same_src_port_rate",
    "dst_host_srv_diff_host_rate",
    "dst_host_serror_rate",
    "dst_host_srv_serror_rate",
    "dst_host_rerror_rate",
    "dst_host_srv_rerror_rate", 
    "attack_type"
]
```

set test data

```python
test_data = pd.read_csv(bronze_dir / "corrected.csv", names=columns)
test_data['attack_category'] = test_data['attack_type'].map(attack_map)
test_data['connection_type'] = np.where(test_data['attack_category'] == 'normal', 'normal', 'attack')
print(f"Corrected data shape: {test_data.shape}")
test_data.head()
```

```python
print(test_data.groupby(by=['attack_category', 'attack_type']).size())
```

print info about the data

set train data

```python
train_data = pd.read_csv(bronze_dir / "kddcup_data_10_percent_corrected.csv", names=columns)
train_data['attack_category'] = train_data['attack_type'].map(attack_map)
train_data['connection_type'] = np.where(train_data['attack_category'] == 'normal', 'normal', 'attack')
print(f"Training data shape: {train_data.shape}")
train_data.head()
```

info about train data

```python
train_data.info()
```

Tutaj też nie ma wartości nullowych

save data to silver directory

```python
silver_dir.mkdir(parents=True, exist_ok=True)

persist_train_data = train_data.drop(columns=['attack_type']).rename(columns={'attack_category':'attack_type'})
persist_test_data = test_data.drop(columns=['attack_type']).rename(columns={'attack_category':'attack_type'})



persist_train_data.to_csv(silver_dir / "train_data.csv", index=False,)
persist_test_data.to_csv(silver_dir / "test_data.csv", index=False)
```
