```python
from pathlib import Path
import py7zr
from shutil import copy2
```

```python
data_dir = Path("../data")

raw_dir = data_dir / "raw"
archive_path = raw_dir / "pliki.7z"
unpacked_dir = raw_dir / "pliki"

bronze_dir = data_dir / "bronze"

suffix = "csv"
```

```python
unpacked_dir.mkdir(parents=True, exist_ok=True)

with py7zr.SevenZipFile(archive_path, mode='r') as archive:
    archive.extractall(path=raw_dir)
```

```python
bronze_dir.mkdir(parents=True, exist_ok=True)

for file in unpacked_dir.iterdir():
    if file.is_file():
        bronze_target = bronze_dir / f"{file.name.replace('.', '_').replace(' ', '_')}.csv"
        copy2(file, bronze_target)
```
